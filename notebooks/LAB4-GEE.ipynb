## Datasets

The datasets being used in the lab are listed below:

- [TIGER: US Census Counties](https://developers.google.com/earth-engine/datasets/catalog/TIGER_2018_Counties)
- [Landsat-9](https://developers.google.com/earth-engine/datasets/catalog/LANDSAT_LC09_C02_T1_L2)
- [Sentinel-2](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR)
- [NAIP](https://developers.google.com/earth-engine/datasets/catalog/USDA_NAIP_DOQQ)

Use [Landsat-9](https://developers.google.com/earth-engine/datasets/catalog/LANDSAT_LC09_C02_T1_L2) data to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
import ee
import geemap
ee.Authenticate()
ee.Initialize(project='projectId')
Map=geemap.Map()

census=ee.FeatureCollection("TIGER/2018/Counties")
N_Knox=census.filter(ee.Filter.eq("NAME", "Knox"))

Tenn_Bounds=(ee.FeatureCollection("TIGER/2018/Counties")
            .filter(ee.Filter.eq("NAME", "Knox"))
            .filter(ee.Filter.eq("STATEFP", "47"))
            )


img=(ee.ImageCollection("LANDSAT/LC09/C02/T1_L2")
            .filterBounds(Tenn_Bounds)
            .filter(ee.Filter.lt('CLOUD_COVER', 30))
            )

landsat=img.median()
landsat_scaled = landsat.multiply(0.0000275).add(-0.2)
landsat_clip = landsat_scaled.clip(Tenn_Bounds)

visualization = {
    "bands": ["SR_B7", "SR_B6", "SR_B4"],
    "min": 0.0,
    "max": 0.3,
    "gamma": 1.5,
}
Tn_style = Tenn_Bounds.style(**{
    "color" : "red",
    "fillColor": "00000000",
    "width" : 1
})

N_Tenne=N_Knox.filter(ee.Filter.eq("STATEFP", "47"))
vis_style = {
    "color": "red",
    "fillColor": "00000000",
}
Map.centerObject(N_Tenne, 10)
Map.addLayer(N_Tenne.style(**vis_style), {}, "Tennessee")

text = "Made by Christine NK"
Map.add_text(text, fontsize=20, position="bottomright")

Map.centerObject(Tenn_Bounds, 10)
Map.addLayer(landsat_clip, visualization, "False(Urban)")
Map.addLayer(Tn_style, {}, "Outline")
Map
```

Use [Sentinel-2](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR) data to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
Map=geemap.Map()

#COPERNICUS/S2_SR dataset has been superseded
img=(ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
        .filterBounds(Tenn_Bounds)
        .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE',10))
        )

img_median=img.median().divide(10000).clip(Tenn_Bounds)
visualization = {
  "min": 0.0,
  "max": 0.3,
  "gamma":1.5,
  "bands": ["B8", "B4", "B3"],
}
out_style=Tenn_Bounds.style(**{
    "color": "red",
    "fillColor": "00000000"
})

Map.addLayer(img_median, visualization, "False(Veg)")
Map.addLayer(out_style, {}, "Outline")
Map.centerObject(Tenn_Bounds, 10)

text = "Made by Christine NK"
Map.add_text(text, fontsize=20, position="bottomright")
Map
```

Use [NAIP](https://developers.google.com/earth-engine/datasets/catalog/USDA_NAIP_DOQQ) imagery to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
Map=geemap.Map()
img=(ee.ImageCollection("USDA/NAIP/DOQQ")
        .filterBounds(Tenn_Bounds)
        )
TrueColor = img.select(['R', 'G', 'B'])
img_median=TrueColor.median().clip(Tenn_Bounds)
visualization = {
  "min": 0,
  "max": 255,
}
out_style=Tenn_Bounds.style(**{
    "color": "red",
    "fillColor": "00000000"
})

Map.addLayer(img_median, visualization, "True")
Map.addLayer(out_style, {}, "Outline")
Map.centerObject(Tenn_Bounds, 10)

text = "Made by Christine NK"
Map.add_text(text, fontsize=20, position="bottomright")
Map
```
